### Part 4: Practical Modeling: Runway Elevation Matrix Analysis (25 Points)
#### Task 4.1: Runway Grid Topography & Gradient Analysis
Analyze 2-D elevation grid with column/row means and hazard thresholds.

```python
# Task 4.1
runway_grid = np.array([
    [14.2, 14.3, 14.5, 14.4, 14.2, 14.1],
    [14.3, 14.5, 14.8, 14.7, 14.4, 14.2],
    [14.4, 14.6, 14.9, 14.8, 14.5, 14.3],
    [14.1, 14.2, 14.4, 14.3, 14.1, 13.9]
])

print("\n--- Part 4 Analysis ---")
print(f"Shape: {runway_grid.shape} | Total Points: {runway_grid.size}")
print(f"Min Elevation: {runway_grid.min():.2f} m | Max Elevation: {runway_grid.max():.2f} m")

col_means = np.mean(runway_grid, axis=0)
row_means = np.mean(runway_grid, axis=1)

print("Longitudinal Station Means (Cols):", np.round(col_means, 2))
print("Lateral Profile Means (Rows):", np.round(row_means, 2))

delta_h = col_means[0] - col_means[-1]
hazard_alert = "[HIGH RIDGE HAZARD] High spot detected!" if np.any(runway_grid > 14.85) else "[NORMAL]"
print(f"Elevation Delta (Stn 0 to 5): {delta_h:+.2f} m | Safety: {hazard_alert}")
```

### Part 5: Challenge Task — Raw Telemetry Stream Parser & Sensor Grid (20 Points)
#### Task 5.1: Radar Telemetry Stream Parsing & Display Table
Parse structured telemetry log strings into a NumPy array, compute metrics, and format a radar summary table.

```python
# Task 5.1
telemetry_log = [
    "AC_ID:THY401,ALT:32000,SPD:450.5,LAT:13.68,LON:100.75",
    "AC_ID:SIA122,ALT:38000,SPD:482.0,LAT:13.72,LON:100.82",
    "AC_ID:BAW009,ALT:29000,SPD:420.2,LAT:13.60,LON:100.68"
]

ids = []
data_rows = []

for line in telemetry_log:
    parts = line.split(",")
    acid = parts[0].split(":")[1]
    alt = float(parts[1].split(":")[1])
    spd = float(parts[2].split(":")[1])
    ids.append(acid)
    data_rows.append([alt, spd])

telemetry_arr = np.array(data_rows)
mean_alt = np.mean(telemetry_arr[:, 0])
max_spd_idx = np.argmax(telemetry_arr[:, 1])

print("\n--- Active Radar Telemetry Summary ---")
print(f"{'Callsign':<10} | {'Altitude (ft)':<13} | {'Speed (kts)':<11}")
print("-" * 40)
for acid, row in zip(ids, telemetry_arr):
    print(f"{acid:<10} | {row[0]:>10,.0f}    | {row[1]:>8.2f}")
print("-" * 40)
print(f"Mean Radar Altitude: {mean_alt:,.2f} ft")
print(f"Fastest Track: {ids[max_spd_idx]} at {telemetry_arr[max_spd_idx, 1]:.2f} kts")
```
