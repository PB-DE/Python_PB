### Part 1: Binary Representation & Basic Concepts (15 Points)
#### Task 1.1: Binary and Integer Conversions (8 Points)
1. Convert 8-bit binary literal `0b10110101` to decimal.
2. Convert altitude `876` to binary using `bin()`.
3. Compute maximum unsigned decimal values for 4, 8, and 16 bits.

```python
# Task 1.1
val_bin = 0b10110101
print("Task 1.1.1 - Decimal value of 0b10110101:", val_bin)  # 181

alt_dec = 876
print("Task 1.1.2 - Binary of altitude 876:", bin(alt_dec))   # '0b1101101100'

max_4bit = 2**4 - 1
max_8bit = 2**8 - 1
max_16bit = 2**16 - 1
print(f"Task 1.1.3 - Max values: 4-bit={max_4bit}, 8-bit={max_8bit}, 16-bit={max_16bit}")
```

#### Task 1.2: Assignment vs. Equality Comparison (7 Points)
Examine equality comparison (`==`) versus assignment (`=`).

```python
# Task 1.2
x = 42
y = 40 + 2
print("Task 1.2 - Equality check (x == y):", x == y)  # True

# Explanation: '==' is a comparison operator that returns a boolean value (True/False).
# In contrast, '=' is an assignment statement in Python; statements cannot be passed
# as expressions inside function arguments like print(), resulting in a SyntaxError.
```

#### Task 2.2: Scientific Formula Translation & Precedence (12 Points)
1. Kinetic Energy ($E = \frac{1}{2}mv^2$)
2. Gravitational Force ($F = G \frac{m_1 m_2}{r^2}$)
3. Operator precedence evaluation

```python
# Task 2.2
# Formula A: Kinetic Energy
m = 1200.0
v = 45.0
E = 0.5 * m * (v ** 2)
print("Task 2.2A - Kinetic Energy:", E, "Joules")

# Formula B: Gravitational Force
G = 6.674e-11
m1 = 5.97e24
m2 = 70.0
r = 6.371e6
F = (G * m1 * m2) / (r ** 2)
print("Task 2.2B - Gravitational Force:", F, "N")

# Formula C: Precedence evaluation
# Step-by-step: 2**3 = 8 -> 3*8 = 24 -> 24/4 = 6.0 -> 6.0//2 = 3.0 -> 10 + 3.0 = 13.0 -> 13.0 - 1 = 12.0
result = 10 + (((3 * (2 ** 3)) / 4) // 2) - 1
print("Task 2.2C - Result:", result, "| Type:", type(result))  # 12.0, <class 'float'>
```

#### Task 3.1: Aircraft Fuel Efficiency Calculator (20 Points)
Prompt for aircraft model, distance flown (NM), and fuel consumed (gallons). Compute efficiency and consumption rate.

### Part 4: Practical Modeling: Moore’s Law Calculator (25 Points)
#### Task 4.1: Moore’s Law Transistor Growth Calculation
Model $N(t) = N_0 \times 2^{\frac{t - t_0}{2}}$ where $t_0 = 1971$ and $N_0 = 2300$.

```python
# Task 4.1
# Example test input: 2023
t0 = 1971
N0 = 2300
target_year = int(input("\nEnter target year: "))
t_elapsed = target_year - t0

transistors = N0 * (2 ** (t_elapsed / 2))
growth_factor = transistors / N0

print(f"Estimated transistors in year {target_year}: {transistors:.2f}")
print(f"Growth factor since 1971: {growth_factor:.2f}x")
```

### Part 5: Challenge Task — Flight Telemetry Descent Speed Check (20 Points)
#### Task 5.1: Descent Speed Safety Verification
Input 3 speed readings in knots, compute min, max, average, and display flight envelope status.

```python
# Task 5.1
# Example test inputs: 145.5, 140.0, 138.5
s1 = float(input("\nEnter speed reading 1 (knots): "))
s2 = float(input("Enter speed reading 2 (knots): "))
s3 = float(input("Enter speed reading 3 (knots): "))

min_speed = min(s1, s2, s3)
max_speed = max(s1, s2, s3)
avg_speed = (s1 + s2 + s3) / 3

print("\n--- Telemetry Summary ---")
print(f"Min Speed: {min_speed:.1f} knots")
print(f"Max Speed: {max_speed:.1f} knots")
print(f"Average Speed: {avg_speed:.2f} knots")

if avg_speed > 160.0:
    print("Status: [WARNING] Approach speed is too fast! Go-around recommended.")
elif avg_speed < 110.0:
    print("Status: [WARNING] Approach speed is too slow! Stall hazard.")
else:
    print("Status: [NORMAL] Approach speed within acceptable landing envelope.")
```
