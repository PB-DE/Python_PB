#### Task 1.2: Boolean Short-Circuit Evaluation (7 Points)
Evaluate short-circuit behavior with `and` / `or` operators.

```python
# Task 1.2
a = 15
b = 0
c = 25
exp1 = (a > 10) and (c < 30)
exp2 = (b != 0) and (c / b > 1)  # Evaluates to False safely without ZeroDivisionError
exp3 = (a == 15) or (b != 0 and c == 25)

print("Task 1.2 Results:", exp1, exp2, exp3)
# Explanation: In Python, 'and' short-circuits. If the first operand (b != 0) is False,
# Python immediately returns False without evaluating (c / b > 1), preventing ZeroDivisionError.
```

### Part 3: Function Architecture & Multiple Return Values (20 Points)
#### Task 3.1 & 3.2: True Airspeed & Mach Number Function (20 Points)
Define `compute_flight_speeds(ias, altitude_ft, temp_c)` returning `(tas, mach)`, call it, and unpack the results.

```python
# Task 3.1 & 3.2
def compute_flight_speeds(ias, altitude_ft, temp_c):
    tas = ias * (1.0 + (altitude_ft / 1000.0) * 0.02)
    speed_of_sound = 38.967 * math.sqrt(temp_c + 273.15)
    mach = tas / speed_of_sound
    return tas, mach

true_airspeed, mach_number = compute_flight_speeds(ias=250.0, altitude_ft=32000.0, temp_c=-45.0)
print(f"Task 3.2 - Calculated TAS: {true_airspeed:.2f} knots")
print(f"Task 3.2 - Calculated Mach: {mach_number:.2f}")
```

### Part 4: Practical Modeling: Runway Assignment & Crosswind Limits (25 Points)
#### Task 4.1: Runway Crosswind & Headwind Limits
Calculate crosswind and headwind components and determine landing clearance.

```python
# Task 4.1
def check_landing_clearance(runway_heading, wind_dir, wind_speed, max_crosswind=15.0):
    diff_deg = wind_dir - runway_heading
    diff_rad = math.radians(diff_deg)
  
    v_cross = wind_speed * abs(math.sin(diff_rad))
    v_head = wind_speed * math.cos(diff_rad)
  
    if v_head < -10.0:
        clearance = "REJECTED: Excessive Tailwind"
    elif v_cross > max_crosswind:
        clearance = "REJECTED: Crosswind Limit Exceeded"
    else:
        clearance = "CLEARED: Safe Landing Conditions"
      
    return clearance, v_cross, v_head

clr, xwind, hwind = check_landing_clearance(90, 130, 22)
print(f"Task 4 - Clearance: {clr} | Crosswind: {xwind:.2f} kts | Headwind: {hwind:.2f} kts")
```

### Part 5: Challenge Task — Stochastic Wind Turbulence Simulation (20 Points)
#### Task 5.1: Monte Carlo Simulation of Wind Shear
Simulate 5,000 approaches with stochastic wind gusts and evaluate alarm probability.

```python
# Task 5.1
random.seed(42)
N_trials = 5000
alarm_count = 0
base_wind = 15.0

for _ in range(N_trials):
    gust = random.randint(-5, 15)
    total_wind = base_wind + gust
    if total_wind > 25.0:
        alarm_count += 1

prob_alarm = (alarm_count / N_trials) * 100.0
assessment = "HIGH CAUTION - Wind shear equipment monitoring mandatory." if prob_alarm > 20.0 else "NORMAL OPERATIONS"

print(f"Task 5 - Approaches: {N_trials}, Alarms: {alarm_count}, Probability: {prob_alarm:.2f}%, Assessment: {assessment}")
```
