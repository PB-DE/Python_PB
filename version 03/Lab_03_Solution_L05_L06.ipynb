### Part 2: Iteration, Range & List Accumulation (20 Points)
#### Task 2.1: Fuel Flow Consumption Accumulator (10 Points)
Calculate sum, average, min, and max fuel flow without using `sum()`.

```python
# Task 2.1
fuel_flows = [120.5, 118.0, 115.2, 112.0, 110.5, 110.0, 108.4, 95.0]

total_flow = 0.0
max_flow = fuel_flows[0]
min_flow = fuel_flows[0]

for flow in fuel_flows:
    total_flow += flow
    if flow > max_flow:
        max_flow = flow
    if flow < min_flow:
        min_flow = flow

avg_flow = total_flow / len(fuel_flows)
print(f"Task 2.1 - Total: {total_flow:.2f} GPH, Avg: {avg_flow:.2f} GPH, Max: {max_flow:.2f}, Min: {min_flow:.2f}")
```

### Part 3: Dictionaries — Key-Value Mapping & Safe Lookups (20 Points)
#### Task 3.1: Airport Fleet Registry (10 Points)
Create, update, add, and iterate over a nested dictionary representing an aircraft fleet manifest.

```python
# Task 3.1
fleet = {
    "TG-401": {"type": "A350-900", "seats": 321, "status": "In-Flight"},
    "TG-402": {"type": "B777-300ER", "seats": 348, "status": "Maintenance"},
    "TG-403": {"type": "B787-9", "seats": 298, "status": "Boarding"},
}

fleet["TG-402"]["status"] = "Available"
fleet["TG-404"] = {"type": "A321neo", "seats": 192, "status": "Standby"}

print("\n--- Active Fleet Manifest ---")
for callsign, info in fleet.items():
    print(f"{callsign:<10} | Type: {info['type']:<12} | Seats: {info['seats']:<4} | Status: {info['status']}")
```

```python
# Task 3.2
target_callsign = "TG-403"
found = fleet.get(target_callsign)
if found:
    print(f"Lookup {target_callsign}: Type {found['type']}, Seats {found['seats']}")
else:
    print("[ALERT] Aircraft callsign not found in active fleet manifest.")
```

### Part 4: Practical Modeling: Weight & Balance Manifest (25 Points)
#### Task 4.1: Cargo CG & Trim Calculation
Calculate total payload weight, moments, center-of-gravity (CG), and verify trim status.

```python
# Task 4.1
cargo_manifest = [
    {"zone": "Zone A (Forward)", "weight_kg": 1420.0, "arm_m": 8.5},
    {"zone": "Zone B (Mid-Fwd)", "weight_kg": 2150.0, "arm_m": 14.2},
    {"zone": "Zone C (Mid-Aft)", "weight_kg": 1890.0, "arm_m": 22.0},
    {"zone": "Zone D (Aft)", "weight_kg": 980.0, "arm_m": 28.5}
]

total_weight = 0.0
total_moment = 0.0

print("\nZone Loading Sheet:")
for item in cargo_manifest:
    weight = item["weight_kg"]
    arm = item["arm_m"]
    moment = weight * arm
    total_weight += weight
    total_moment += moment
    print(f"- {item['zone']}: {weight:.1f} kg | Arm: {arm:.1f} m | Moment: {moment:.1f} kg-m")

cg_location = total_moment / total_weight
trim_status = "[SAFE TRIM NORMAL]" if (14.0 <= cg_location <= 18.0) else "[UNSAFE CG WARNING]"

print(f"\nTotal Cargo Weight: {total_weight:.2f} kg")
print(f"Total Moment: {total_moment:.2f} kg-m")
print(f"Calculated CG Location: {cg_location:.2f} meters")
print(f"Trim Status: {trim_status}")
```

### Part 5: Challenge Task — Flight Waypoint Analyzer with Sets (20 Points)
#### Task 5.1: Navigation Corridors & Set Operations
Perform set intersection (`&`), difference (`-`), union (`|`), and subset verification (`issubset()`).

```python
# Task 5.1
flight_A_waypoints = ["BKK", "LIMLA", "BETNO", "SALAX", "DOSEN", "CONOR", "HKG"]
flight_B_waypoints = ["CNX", "BETNO", "SURIN", "DOSEN", "CANTO", "HKG", "TPE"]

set_A = set(flight_A_waypoints)
set_B = set(flight_B_waypoints)

shared = set_A & set_B
unique_A = set_A - set_B
all_waypoints = set_A | set_B

check_pair = {"BETNO", "DOSEN"}
in_A = check_pair.issubset(set_A)
in_B = check_pair.issubset(set_B)

print("\nTask 5 Results:")
print("Shared Intersecting Waypoints:", shared)
print("Waypoints Unique to Flight A:", unique_A)
print(f"Total Consolidated Waypoints: {len(all_waypoints)} waypoints")
print(f"Waypoint Pair {check_pair} in Flight A: {in_A}")
print(f"Waypoint Pair {check_pair} in Flight B: {in_B}")
```
